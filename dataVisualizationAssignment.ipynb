###  5. Generate a synthetic dataset with correlated features. Visualize the correlation matrix of a dataset using a heatmap.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(42)

n_samples = 1000

correlation_matrix = np.array([[1.0, 0.8, 0.6],
                                [0.8, 1.0, 0.5],
                                [0.6, 0.5, 1.0]])

L = np.linalg.cholesky(correlation_matrix)
random_data = np.random.normal(size=(n_samples, correlation_matrix.shape[0]))
synthetic_data = np.dot(random_data, L)

df = pd.DataFrame(synthetic_data, columns=['Feature 1', 'Feature 2', 'Feature 3'])

corr = df.corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f", square=True, cbar_kws={"shrink": .8})

plt.title('Correlation Matrix Heatmap')
plt.xlabel('Features')
plt.ylabel('Features')

plt.show()
````

### 4. Using the given x and y data, generate a 3D surface plot to visualize the function z = np.sin(np.sqrt(x**2 + y**2)).

```x = np.linspace(-5, 5, 100) y = np.linspace(-5, 5, 100) x, y = np.meshgrid(x, y) z = np.sin(np.sqrt(x**2 + y**2))data = {'X': x.flatten(),'Y': y.flatten(),'Z': z.flatten()} df = pd.DataFrame(data)```

````python
import numpy as np
import pandas as pd
import plotly.graph_objects as go

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
x, y = np.meshgrid(x, y)

z = np.sin(np.sqrt(x**2 + y**2))

data = {
    'X': x.flatten(),
    'Y': y.flatten(),
    'Z': z.flatten()
}
df = pd.DataFrame(data)

fig = go.Figure(data=[go.Surface(z=z, x=x, y=y)])

fig.update_layout(title='3D Surface Plot of z = sin(sqrt(x^2 + y^2))',
                  scene=dict(
                      xaxis_title='X-axis',
                      yaxis_title='Y-axis',
                      zaxis_title='Z-axis'
                  ))

fig.show()
````

# **Bokeh**

###  1.Create a Bokeh plot displaying a sine wave. Set x-values from 0 to 10 and y-values as the sine of x.






````python
import numpy as np
from bokeh.plotting import figure, output_notebook, show

output_notebook()

x = np.linspace(0, 10, 100)
y = np.sin(x)

p = figure(title="Sine Wave", x_axis_label='x', y_axis_label='sin(x)', width=800, height=400)

p.line(x, y, line_width=2, line_color='blue')

show(p)
````

###  2.Create a Bokeh scatter plot using randomly generated x and y values. Use different sizes and colors for the markers based on the 'sizes' and 'colors' columns.

````python
import numpy as np
from bokeh.plotting import figure, output_notebook, show
from bokeh.models import ColumnDataSource

output_notebook()

np.random.seed(42)
N = 100
x = np.random.rand(N) * 10
y = np.random.rand(N) * 10
sizes = np.random.randint(5, 50, size=N)
colors = np.random.choice(['red', 'green', 'blue', 'orange', 'purple'], size=N)

source = ColumnDataSource(data=dict(x=x, y=y, sizes=sizes, colors=colors))

p = figure(title="Random Scatter Plot", x_axis_label='X-axis', y_axis_label='Y-axis', width=800, height=400)

p.scatter('x', 'y', size='sizes', color='colors', alpha=0.6, source=source)

show(p)
````

###  3. Generate a Bokeh bar chart representing the counts of different fruits using the following dataset.

````python
from bokeh.plotting import figure, show, output_notebook

fruits = ['Apples', 'Oranges', 'Bananas', 'Pears']
counts = [20, 25, 30, 35]

output_notebook()

p = figure(x_range=fruits, title="Fruit Counts",
           x_axis_label='Fruits', y_axis_label='Counts',
           height=400, width=600)

p.vbar(x=fruits, top=counts, width=0.5)

show(p)
````

 ### 4. Create a Bokeh histogram to visualize the distribution of the given data.

 ```data_hist = np.random.randn(1000)hist, edges = np.histogram(data_hist, bins=30)```

````python
import numpy as np
from bokeh.plotting import figure, show, output_notebook

data_hist = np.random.randn(1000)

hist, edges = np.histogram(data_hist, bins=30)

output_notebook()

p = figure(title="Histogram of Randomly Generated Data",
           x_axis_label='Value',
           y_axis_label='Frequency',
           height=400,
           width=600)

p.quad(top=hist, bottom=0, left=edges[:-1], right=edges[1:],
        fill_color="skyblue", line_color="white")

show(p)
````

###  5. Create a Bokeh heatmap using the provided dataset.

 ```data_heatmap = np.random.rand(10, 10)x = np.linspace(0, 1, 10) y = np.linspace(0, 1, 10) xx, yy = np.meshgrid(x, y)```

````python
import numpy as np
import pandas as pd
from bokeh.plotting import figure, show, output_notebook
from bokeh.transform import linear_cmap
from bokeh.models import ColorBar, LinearColorMapper

data_heatmap = np.random.rand(10, 10)
x = np.linspace(0, 1, 10)
y = np.linspace(0, 1, 10)

xx, yy = np.meshgrid(x, y)

data = {
    'x': xx.flatten(),
    'y': yy.flatten(),
    'value': data_heatmap.flatten()
}

df = pd.DataFrame(data)

output_notebook()

mapper = LinearColorMapper(palette="Viridis256", low=df['value'].min(), high=df['value'].max())

p = figure(title="Heatmap of Random Data",
           x_axis_label='X',
           y_axis_label='Y',
           tools="",
           width=600,
           height=600)

p.rect(x='x', y='y', width=0.1, height=0.1, source=df,
        line_color=None, fill_color=linear_cmap('value', palette="Viridis256", low=df['value'].min(), high=df['value'].max()))

color_bar = ColorBar(color_mapper=mapper, location=(0, 0))
p.add_layout(color_bar, 'right')

show(p)
````
