### 1  Generate a list of 100 integers containing values between 90 to 130 and store it in the variable `int_list`. After generating the list, find the following:

 - (i) Write a Python function to calculate the mean of a given list of numbers.
 Create a function to find the median of a list of numbers.
 - (ii) Develop a program to compute the mode of a list of integers.
 - (iii) Implement a function to calculate the weighted mean of a list of values and their corresponding weights.
 - (iv) Write a Python function to find the geometric mean of a list of positive numbers.
 - (v) Create a program to calculate the harmonic mean of a list of values.
 - (vi) Build a function to determine the midrange of a list of numbers (average of the minimum and maximum).
 - (vii) Implement a Python program to find the trimmed mean of a list, excluding a certain percentage of
outliers.

````python
import numpy as np
from scipy import stats

# Generate a list of 100 integers between 90 and 130
np.random.seed(42)  # For reproducibility
int_list = np.random.randint(90, 131, size=100).tolist()

# (i) Function to calculate the mean of a given list of numbers
def calculate_mean(numbers):
    return sum(numbers) / len(numbers)

# Function to find the median of a list of numbers
def calculate_median(numbers):
    sorted_numbers = sorted(numbers)
    n = len(sorted_numbers)
    mid = n // 2
    if n % 2 == 0:
        return (sorted_numbers[mid - 1] + sorted_numbers[mid]) / 2
    else:
        return sorted_numbers[mid]

# (ii) Function to compute the mode of a list of integers
def calculate_mode(numbers):
    mode_result = stats.mode(numbers)
    return mode_result.mode, mode_result.count

# (iii) Function to calculate the weighted mean of a list of values and their corresponding weights
def calculate_weighted_mean(values, weights):
    return sum(value * weight for value, weight in zip(values, weights)) / sum(weights)

# (iv) Function to find the geometric mean of a list of positive numbers
def calculate_geometric_mean(numbers):
    return stats.gmean(numbers)

# (v) Function to calculate the harmonic mean of a list of values
def calculate_harmonic_mean(numbers):
    return stats.hmean(numbers)

# (vi) Function to determine the midrange of a list of numbers
def calculate_midrange(numbers):
    return (min(numbers) + max(numbers)) / 2

# (vii) Function to find the trimmed mean of a list, excluding a certain percentage of outliers
def calculate_trimmed_mean(numbers, proportion_to_trim):
    return stats.trim_mean(numbers, proportion_to_trim)

# Example usage:
print("Generated Integer List:", int_list)
print("i) a) Mean:", calculate_mean(int_list))
print("i) b) Median:", calculate_median(int_list))
mode_value, mode_count = calculate_mode(int_list)
print("ii) Mode:", mode_value, "Count:", mode_count)

# For weighted mean example:
weights = np.random.rand(100)  # Random weights for demonstration
print("iii)  Weighted Mean:", calculate_weighted_mean(int_list, weights))

# Geometric Mean
print("iv) Geometric Mean:", calculate_geometric_mean(int_list))

# Harmonic Mean
print("v) Harmonic Mean:", calculate_harmonic_mean(int_list))

# Midrange
print("vi) Midrange:", calculate_midrange(int_list))

# Trimmed Mean (e.g., trimming 10% from both ends)
print("vii) Trimmed Mean (10%):", calculate_trimmed_mean(int_list, proportion_to_trim=0.1))
````

### 3 Write a Python class representing a discrete random variable with methods to calculate its expected value and variance.

````python
class DiscreteRandomVariable:
    def __init__(self, values, probabilities):

        if len(values) != len(probabilities):
            raise ValueError("Values and probabilities must have the same length.")

        if not np.isclose(sum(probabilities), 1):
            raise ValueError("Probabilities must sum to 1.")

        self.values = values
        self.probabilities = probabilities

    def expected_value(self):

        return sum(value * prob for value, prob in zip(self.values, self.probabilities))

    def variance(self):

        mean = self.expected_value()
        return sum(prob * (value - mean) ** 2 for value, prob in zip(self.values, self.probabilities))

if __name__ == "__main__":
    # Define values and their corresponding probabilities
    values = [0, 1, 2, 3]
    probabilities = [0.1, 0.3, 0.4, 0.2]  # Must sum to 1

    # Create a DiscreteRandomVariable instance
    random_variable = DiscreteRandomVariable(values, probabilities)

    # Calculate expected value and variance
    expected_value = random_variable.expected_value()
    variance = random_variable.variance()

    print("Expected Value:", expected_value)
    print("Variance:", variance)
````

### 4  Implement a program to simulate the rolling of a fair six-sided die and calculate the expected value and variance of the outcomes.

````python
import numpy as np

class FairDie:
    def __init__(self):
        self.values = [1, 2, 3, 4, 5, 6]  # Possible outcomes of a six-sided die
        self.probabilities = [1/6] * 6     # Equal probability for each outcome

    def expected_value(self):
        """
        Calculates the expected value (mean) of rolling the die.

        :return: The expected value.
        """
        return sum(value * prob for value, prob in zip(self.values, self.probabilities))

    def variance(self):
        """
        Calculates the variance of rolling the die.

        :return: The variance.
        """
        mean = self.expected_value()
        return sum(prob * (value - mean) ** 2 for value, prob in zip(self.values, self.probabilities))

    def roll(self, num_rolls):
        """
        Simulates rolling the die a specified number of times.

        :param num_rolls: The number of times to roll the die.
        :return: A list of outcomes from rolling the die.
        """
        return np.random.choice(self.values, size=num_rolls, p=self.probabilities)

if __name__ == "__main__":
    die = FairDie()

    # Calculate expected value and variance
    expected_value = die.expected_value()
    variance = die.variance()

    print("Expected Value of Die Rolls:", expected_value)
    print("Variance of Die Rolls:", variance)

    # Simulate rolling the die 1000 times
    num_rolls = 1000
    outcomes = die.roll(num_rolls)

    # Display some statistics from the simulation
    print("Outcomes from Rolling the Die:", outcomes)
    print("Mean of Outcomes from Simulation:", np.mean(outcomes))
    print("Variance of Outcomes from Simulation:", np.var(outcomes))
````

### 5  Create a Python function to generate random samples from a given probability distribution (e.g., binomial, Poisson) and calculate their mean and variance.

````python
import numpy as np

def generate_binomial_samples(n, p, size):
    """
    Generate random samples from a binomial distribution.

    :param n: Number of trials.
    :param p: Probability of success on each trial.
    :param size: Number of random samples to generate.
    :return: A tuple containing the samples, mean, and variance.
    """
    samples = np.random.binomial(n, p, size)
    mean = np.mean(samples)
    variance = np.var(samples)
    return samples, mean, variance

def generate_poisson_samples(lam, size):
    """
    Generate random samples from a Poisson distribution.

    :param lam: The expected number of events (lambda).
    :param size: Number of random samples to generate.
    :return: A tuple containing the samples, mean, and variance.
    """
    samples = np.random.poisson(lam, size)
    mean = np.mean(samples)
    variance = np.var(samples)
    return samples, mean, variance

if __name__ == "__main__":
    # Parameters for the binomial distribution
    n = 10   # Number of trials
    p = 0.5  # Probability of success
    size = 1000  # Number of samples

    # Generate binomial samples
    binomial_samples, binomial_mean, binomial_variance = generate_binomial_samples(n, p, size)

    print("Binomial Samples:", binomial_samples[:10])  # Print first 10 samples
    print("Mean of Binomial Samples:", binomial_mean)
    print("Variance of Binomial Samples:", binomial_variance)

    # Parameters for the Poisson distribution
    lam = 5  # Expected number of events (lambda)

    # Generate Poisson samples
    poisson_samples, poisson_mean, poisson_variance = generate_poisson_samples(lam, size)

    print("\nPoisson Samples:", poisson_samples[:10])  # Print first 10 samples
    print("Mean of Poisson Samples:", poisson_mean)
    print("Variance of Poisson Samples:", poisson_variance)
````

### 6  Write a Python script to generate random numbers from a Gaussian (normal) distribution and compute the mean, variance, and standard deviation of the samples.

````python
import numpy as np

def generate_gaussian_samples(mean, std_dev, size):
    """
    Generate random samples from a Gaussian (normal) distribution.

    :param mean: Mean of the normal distribution.
    :param std_dev: Standard deviation of the normal distribution.
    :param size: Number of random samples to generate.
    :return: A tuple containing the samples, mean, variance, and standard deviation.
    """
    samples = np.random.normal(mean, std_dev, size)
    sample_mean = np.mean(samples)
    sample_variance = np.var(samples)
    sample_std_dev = np.std(samples)

    return samples, sample_mean, sample_variance, sample_std_dev

if __name__ == "__main__":
    # Parameters for the Gaussian distribution
    mean = 0          # Mean of the distribution
    std_dev = 1      # Standard deviation of the distribution
    size = 1000      # Number of samples to generate

    # Generate Gaussian samples
    gaussian_samples, sample_mean, sample_variance, sample_std_dev = generate_gaussian_samples(mean, std_dev, size)

    # Output results
    print("First 10 Gaussian Samples:", gaussian_samples[:10])  # Print first 10 samples
    print("Mean of Samples:", sample_mean)
    print("Variance of Samples:", sample_variance)
    print("Standard Deviation of Samples:", sample_std_dev)
````

### 8  Write a Python function to calculate the probability density function (PDF) of a continuous random variable for a given normal distribution.

````python
import numpy as np

def normal_pdf(x, mu=0, sigma=1):
    """
    Calculate the probability density function (PDF) of a normal distribution.

    :param x: The point(s) at which to evaluate the PDF.
    :param mu: Mean of the normal distribution (default is 0).
    :param sigma: Standard deviation of the normal distribution (default is 1).
    :return: The value(s) of the PDF at point(s) x.
    """
    # Ensure sigma is positive
    if sigma <= 0:
        raise ValueError("Standard deviation must be positive.")

    # Calculate the PDF using the formula
    pdf_value = (1 / (sigma * np.sqrt(2 * np.pi))) * np.exp(-((x - mu) ** 2) / (2 * sigma ** 2))

    return pdf_value

if __name__ == "__main__":
    # Parameters for the normal distribution
    mu = 0       # Mean
    sigma = 1    # Standard deviation

    # Points at which to evaluate the PDF
    x_values = np.linspace(-3, 3, 100)

    # Calculate PDF values for these points
    pdf_values = normal_pdf(x_values, mu, sigma)

    # Print some results
    print("X values:", x_values[:10])  # Print first 10 x values
    print("PDF values:", pdf_values[:10])  # Print first 10 PDF values
````

### 9  Create a program to calculate the cumulative distribution function (CDF) of exponential distribution.


````python
import numpy as np
import matplotlib.pyplot as plt

def exponential_cdf(x, lambd):
    """
    Calculate the cumulative distribution function (CDF) of an exponential distribution.

    :param x: The point(s) at which to evaluate the CDF.
    :param lambd: The rate parameter (lambda) of the exponential distribution.
    :return: The value(s) of the CDF at point(s) x.
    """
    if lambd <= 0:
        raise ValueError("Rate parameter (lambda) must be positive.")

    return 1 - np.exp(-lambd * x)

if __name__ == "__main__":
    # Parameters for the exponential distribution
    lambd = 1.0  # Rate parameter (mean = 1/lambda)

    # Generate x values from 0 to 5
    x_values = np.linspace(0, 5, 100)

    # Calculate CDF values for these points
    cdf_values = exponential_cdf(x_values, lambd)

    # Print some results
    print("X values:", x_values[:10])  # Print first 10 x values
    print("CDF values:", cdf_values[:10])  # Print first 10 CDF values

    # Plotting the CDF
    plt.figure(figsize=(8, 5))
    plt.plot(x_values, cdf_values, label='CDF', color='blue')
    plt.title('Cumulative Distribution Function (CDF) of Exponential Distribution')
    plt.xlabel('x')
    plt.ylabel('CDF')
    plt.grid()
    plt.legend()
    plt.show()
````

### 10 Write a Python function to calculate the probability mass function (PMF) of Poisson distribution.

````python
import numpy as np
from scipy.special import factorial

def poisson_pmf(k, lambd):
    """
    Calculate the probability mass function (PMF) of a Poisson distribution.

    :param k: The number of occurrences (non-negative integer).
    :param lambd: The average rate of occurrence (lambda).
    :return: The PMF value at k.
    """
    if k < 0 or not isinstance(k, int):
        raise ValueError("k must be a non-negative integer.")
    if lambd < 0:
        raise ValueError("Lambda must be non-negative.")

    return (np.exp(-lambd) * (lambd ** k)) / factorial(k)

if __name__ == "__main__":
    # Parameters for the Poisson distribution
    lambd = 3.0  # Average rate of occurrence

    # Calculate PMF for k values from 0 to 10
    k_values = range(11)
    pmf_values = [poisson_pmf(k, lambd) for k in k_values]

    # Print results
    for k, pmf in zip(k_values, pmf_values):
        print(f"PMF(k={k}): {pmf:.4f}")
````

### 11  A company wants to test if a new website layout leads to a higher conversion rate (percentage of visitors who make a purchase). They collect data from the old and new layouts to compare.To generate the data use the following command:
 ```python
 import numpy as np
 # 50 purchases out of 1000 visitors
 old_layout = np.array([1] * 50 + [0] * 950)
 # 70 purchases out of 1000 visitors  
new_layout = np.array([1] * 70 + [0] * 930)
 ```
###Apply z-test to find which layout is successful.

````python
import numpy as np
import scipy.stats as stats

old_layout = np.array([1] * 50 + [0] * 950)
# 70 purchases out of 1000 visitors for new layout
new_layout = np.array([1] * 70 + [0] * 930)

# Calculate the number of successes and sample sizes
x1 = np.sum(old_layout)  # Number of purchases in old layout
n1 = len(old_layout)      # Total visitors in old layout

x2 = np.sum(new_layout)   # Number of purchases in new layout
n2 = len(new_layout)       # Total visitors in new layout

# Calculate conversion rates
p1 = x1 / n1  # Conversion rate for old layout
p2 = x2 / n2  # Conversion rate for new layout

# Calculate pooled probability
p = (x1 + x2) / (n1 + n2)

# Calculate the Z-score
z = (p1 - p2) / np.sqrt(p * (1 - p) * (1/n1 + 1/n2))

# Calculate the p-value for a two-tailed test
p_value = stats.norm.sf(abs(z)) * 2  # Multiply by 2 for two-tailed

# Print results
print(f"Conversion Rate Old Layout: {p1:.4f}")
print(f"Conversion Rate New Layout: {p2:.4f}")
print(f"Z-Score: {z:.4f}")
print(f"P-Value: {p_value:.4f}")

# Interpretation of results
alpha = 0.05  # Significance level
if p_value < alpha:
    print("Reject the null hypothesis: The new layout has a significantly higher conversion rate.")
else:
    print("Fail to reject the null hypothesis: No significant difference in conversion rates.")
````

### 12  A tutoring service claims that its program improves students' exam scores. A sample of students who participated in the program was taken, and their scores before and after the program were recorded.

 Use the below code to generate samples of respective arrays of marks:
 ```python
 before_program = np.array([75, 80, 85, 70, 90, 78, 92, 88, 82, 87])
 after_program = np.array([80, 85, 90, 80, 92, 80, 95, 90, 85, 88])
 ```
 ### Use z-test to find if the claims made by tutor are true or false.

````python
import numpy as np
import scipy.stats as stats

# Generate sample data
before_program = np.array([75, 80, 85, 70, 90, 78, 92, 88, 82, 87])
after_program = np.array([80, 85, 90, 80, 92, 80, 95, 90, 85, 88])

# Calculate differences
differences = after_program - before_program

# Calculate mean and standard deviation of differences
mean_difference = np.mean(differences)
std_difference = np.std(differences, ddof=1)  # Sample standard deviation
n = len(differences)

# Calculate Z-score
z_score = mean_difference / (std_difference / np.sqrt(n))

# Calculate p-value for one-tailed test
p_value = stats.norm.sf(z_score)  # sf is the survival function (1 - CDF)

# Print results
print(f"Mean Difference: {mean_difference:.4f}")
print(f"Standard Deviation of Differences: {std_difference:.4f}")
print(f"Z-Score: {z_score:.4f}")
print(f"P-Value: {p_value:.4f}")

# Interpretation of results
alpha = 0.05  # Significance level
if p_value < alpha:
    print("Reject the null hypothesis: The tutoring program significantly improves exam scores.")
else:
    print("Fail to reject the null hypothesis: No significant improvement in exam scores.")
````

### 15  A company is testing two different website layouts to see which one leads to higher click-through rates. Write a Python function to perform an A/B test analysis, including calculating the t-statistic, degrees of freedom, and p-value.

 Use the following data:
 ```python
 layout_a_clicks = [28, 32, 33, 29, 31, 34, 30, 35, 36, 37]
 layout_b_clicks = [40, 41, 38, 42, 39, 44, 43, 41, 45, 47]

````python
import numpy as np
from scipy import stats

def ab_test(layout_a_clicks, layout_b_clicks):
    # Convert lists to numpy arrays for calculations
    a = np.array(layout_a_clicks)
    b = np.array(layout_b_clicks)

    # Calculate means and standard deviations
    mean_a = np.mean(a)
    mean_b = np.mean(b)
    std_a = np.std(a, ddof=1)  # Sample standard deviation
    std_b = np.std(b, ddof=1)  # Sample standard deviation
    n_a = len(a)
    n_b = len(b)

    # Calculate t-statistic
    t_statistic = (mean_a - mean_b) / np.sqrt((std_a**2 / n_a) + (std_b**2 / n_b))

    # Calculate degrees of freedom
    df_numerator = ((std_a**2 / n_a) + (std_b**2 / n_b))**2
    df_denominator = ((std_a**2 / n_a)**2 / (n_a - 1)) + ((std_b**2 / n_b)**2 / (n_b - 1))
    degrees_of_freedom = df_numerator / df_denominator

    # Calculate p-value for two-tailed test
    p_value = stats.t.sf(np.abs(t_statistic), df=degrees_of_freedom) * 2  # Two-tailed test

    # Print results
    print(f"Layout A Mean Clicks: {mean_a:.4f}")
    print(f"Layout B Mean Clicks: {mean_b:.4f}")
    print(f"T-Statistic: {t_statistic:.4f}")
    print(f"Degrees of Freedom: {degrees_of_freedom:.4f}")
    print(f"P-Value: {p_value:.4f}")

    # Interpretation of results
    alpha = 0.05  # Significance level
    if p_value < alpha:
        print("Reject the null hypothesis: There is a significant difference in click-through rates.")
    else:
        print("Fail to reject the null hypothesis: No significant difference in click-through rates.")

# Example usage with provided data
layout_a_clicks = [28, 32, 33, 29, 31, 34, 30, 35, 36, 37]
layout_b_clicks = [40, 41, 38, 42, 39, 44, 43, 41, 45, 47]

ab_test(layout_a_clicks, layout_b_clicks)
````

### 21 A political analyst wants to determine if there is a significant association between age groups and voter preferences (Candidate A or Candidate B). They collect data from a sample of 500 voters and classify them into different age groups and candidate preferences. Perform a Chi-Square test to determine if there is a significant association between age groups and voter preferences.

 Use the below code to generate data:
 ```python
 np.random.seed(0)
 age_groups = np.random.choice(['18-30', '31-50', '51+', '51+'], size=30)
 voter_preferences = np.random.choice(['Candidate A', 'Candidate B'], size=30)
 ```

````python
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency

# Generate synthetic data
np.random.seed(0)
age_groups = np.random.choice(['18-30', '31-50', '51+'], size=500)
voter_preferences = np.random.choice(['Candidate A', 'Candidate B'], size=500)

# Create a DataFrame for better visualization
data = pd.DataFrame({'Age Group': age_groups, 'Voter Preference': voter_preferences})

# Create a contingency table
contingency_table = pd.crosstab(data['Age Group'], data['Voter Preference'])

# Perform Chi-Square test
chi2_stat, p_value, dof, expected = chi2_contingency(contingency_table)

# Print results
print("Contingency Table:")
print(contingency_table)
print(f"\nChi-Squared Statistic: {chi2_stat:.4f}")
print(f"P-Value: {p_value:.4f}")
print(f"Degrees of Freedom: {dof}")
print("\nExpected Frequencies:")
print(expected)

# Interpretation of results
alpha = 0.05  # Significance level
if p_value < alpha:
    print("Reject the null hypothesis: There is a significant association between age groups and voter preferences.")
else:
    print("Fail to reject the null hypothesis: No significant association between age groups and voter preferences.")
````
