### 3  Explain the following terms
- The difference in np.array, np.asarray and np.asanyarray
- The difference between Deep copy and shallow copy


### Differences Between `np.array`, `np.asarray`, and `np.asanyarray`

i. **`np.array`**:
   - **Purpose**: Creates a new NumPy array from a given input, copying the data by default.
   - **Behavior**: It allows you to specify the desired data type (`dtype`) and can create arrays from lists, tuples, or other sequences.
   - **Example**:
     ```python
     import numpy as np
     arr1 = np.array([1, 2, 3], dtype=float)
     print(arr1)  # Output: [1. 2. 3.]
     ```

ii. **`np.asarray`**:
   - **Purpose**: Converts the input to an array but does not copy the data if the input is already an array of the same type.
   - **Behavior**: It is more efficient than `np.array` when you want to ensure that the output is an array without unnecessary copying.
   - **Example**:
     ```python
     arr2 = np.asarray(arr1)  # arr1 is already an array
     print(arr2 is arr1)  # Output: True (no copy made)
     ```

iii. **`np.asanyarray`**:
   - **Purpose**: Similar to `np.asarray`, but it allows for subclasses of arrays (like masked arrays) to be returned as they are.
   - **Behavior**: It is useful when you want to preserve subclasses (e.g., `np.ma.masked_array`) while converting inputs to an array.
   - **Example**:
     ```python
     arr3 = np.asanyarray(arr1)  # arr1 is a regular array
     print(arr3 is arr1)  # Output: True (no copy made)
     ```



### Differences Between Deep Copy and Shallow Copy

 **Shallow Copy**:
   - **Definition**: A shallow copy creates a new object but inserts references into it to the objects found in the original. Thus, changes made to mutable objects in the copied object will reflect in the original object.
   - **Example**:
     ```python
     import copy

     original_list = [[1, 2, 3], [4, 5, 6]]
     shallow_copied_list = copy.copy(original_list)

     shallow_copied_list[0][0] = 'X'
     print(original_list)  # Output: [['X', 2, 3], [4, 5, 6]] (original is affected)
     ```

 **Deep Copy**:
   - **Definition**: A deep copy creates a new object and recursively adds copies of nested objects found in the original. Changes made to mutable objects in the copied object do not affect the original object.
   - **Example**:
     ```python
     import copy

     original_list = [[1, 2, 3], [4, 5, 6]]
     deep_copied_list = copy.deepcopy(original_list)

     deep_copied_list[0][0] = 'X'
     print(original_list)  # Output: [[1, 2, 3], [4, 5, 6]] (original remains unchanged)
     ```

### 7  Clean and transform the 'Phone' column in the sample dataset to remove non-numeric characters and convert it to a numeric data type. Also display the table attributes and data types of each column.

### 14 Create a time-series dataset in a Pandas DataFrame with columns: 'Date', 'Temperature', 'Humidity' and Perform the following tasks using Matplotlib:
- a) Plot the 'Temperature' and 'Humidity' on the same plot with different y-axes (left y-axis for 'Temperature' and
 right y-axis for 'Humidity').
- b) Label the x-axis as 'Date'.
- c) Set the title of the plot as 'Temperature and Humidity Over Time'.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

date_range = pd.date_range(start='2023-01-01', periods=100, freq='D')

# a)
temperature = np.random.uniform(15, 30, size=len(date_range))

humidity = np.random.uniform(30, 90, size=len(date_range))

data = {
    'Date': date_range,
    'Temperature': temperature,
    'Humidity': humidity
}
df = pd.DataFrame(data)

fig, ax1 = plt.subplots()

# b)
ax1.set_xlabel('Date')
ax1.set_ylabel('Temperature (°C)', color='tab:red')
ax1.plot(df['Date'], df['Temperature'], color='red', label='Temperature')
ax1.tick_params(axis='y', labelcolor='tab:red')

ax2 = ax1.twinx()
ax2.set_ylabel('Humidity (%)', color='tab:blue')
ax2.plot(df['Date'], df['Humidity'], color='blue', label='Humidity')
ax2.tick_params(axis='y', labelcolor='tab:blue')

# c) : Set the title of the plot
plt.title('Temperature and Humidity Over Time')

fig.tight_layout()
plt.show()
````

### 15  Create a NumPy array data containing 1000 samples from a normal distribution. Perform the following tasks using Matplotlib:
- a) Plot a histogram of the data with 30 bins.
- b) Overlay a line plot representing the normal distribution's probability density function (PDF).
- c) Label the x-axis as 'Value' and the y-axis as 'Frequency/Probability'.
- d) Set the title of the plot as 'Histogram with PDF Overlay'.

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

data = np.random.normal(loc=0, scale=1, size=1000)  # Mean = 0, Standard deviation = 1

# a) : Plot a histogram of the data with 30 bins
plt.hist(data, bins=30, density=True, alpha=0.6, color='g', label='Histogram')

# b) : Overlay the PDF of the normal distribution
# Calculate the PDF values for a range of x values
xmin, xmax = plt.xlim()  # Get the limits of x-axis
x = np.linspace(xmin, xmax, 100)
p = norm.pdf(x, loc=0, scale=1)  # PDF for a normal distribution with mean=0 and std=1
plt.plot(x, p, 'k', linewidth=2, label='Normal Distribution PDF')

# c) : Label the axes
plt.xlabel('Value')
plt.ylabel('Frequency/Probability')

# d) : Set the title of the plot
plt.title('Histogram with PDF Overlay')

plt.legend()

plt.show()
````

### 17 Create a Seaborn scatter plot of two random arrays, color points based on their position relative to the origin (quadrants), add a legend, label the axes, and set the title as 'Quadrant-wise Scatter Plot'.

````python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

np.random.seed(0)
x = np.random.rand(100) * 10 - 5
y = np.random.rand(100) * 10 - 5

quadrants = []
for i in range(len(x)):
    if x[i] >= 0 and y[i] >= 0:
        quadrants.append('Quadrant I')
    elif x[i] < 0 and y[i] >= 0:
        quadrants.append('Quadrant II')
    elif x[i] < 0 and y[i] < 0:
        quadrants.append('Quadrant III')
    else:
        quadrants.append('Quadrant IV')

data = pd.DataFrame({'X': x, 'Y': y, 'Quadrant': quadrants})

plt.figure(figsize=(10, 6))
sns.scatterplot(data=data, x='X', y='Y', hue='Quadrant', style='Quadrant', palette='Set1', s=100)

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Quadrant-wise Scatter Plot')

plt.legend(title='Quadrants')

plt.axhline(0, color='black', lw=1)
plt.axvline(0, color='black', lw=1)
plt.grid()
plt.show()
````

### 18 With Bokeh, plot a line chart of a sine wave function, add grid lines, label the axes, and set the title as 'Sine Wave Function'

````python
from bokeh.plotting import figure, show
from bokeh.io import output_notebook
import numpy as np

output_notebook()

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

p = figure(title="Sine Wave Function", x_axis_label='X', y_axis_label='Y', width=800, height=400)

p.line(x, y, line_width=2, color='blue', legend_label='sin(x)')

p.grid.grid_line_alpha = 0.3

show(p)
````

### 19 Using Bokeh, generate a bar chart of randomly generated categorical data, color bars based on their values, add hover tooltips to display exact values, label the axes, and set the title as 'Random Categorical Bar Chart'

````python
import numpy as np
import pandas as pd
from bokeh.plotting import figure, show
from bokeh.io import output_notebook
from bokeh.models import ColumnDataSource, HoverTool

output_notebook()

categories = [f'Category {i}' for i in range(10)]
values = np.random.randint(1, 100, size=len(categories))

data = pd.DataFrame({'Category': categories, 'Values': values})

source = ColumnDataSource(data=data)

p = figure(x_range=categories, title="Random Categorical Bar Chart",
           toolbar_location=None, tools="")

p.vbar(x='Category', top='Values', source=source, width=0.9,
        line_color='white', fill_color='blue')

hover = HoverTool()
hover.tooltips = [("Category", "@Category"), ("Value", "@Values")]
p.add_tools(hover)

p.xaxis.axis_label = "Categories"
p.yaxis.axis_label = "Values"
p.y_range.start = 0

p.grid.grid_line_color = "gray"

show(p)
````

### 20 Using Plotly, create a basic line plot of a randomly generated dataset, label the axes, and set the title as 'Simple Line Plot'

````python
import numpy as np
import plotly.graph_objs as go
import plotly.offline as pyo

x = np.linspace(0, 10, 100)
y = np.random.rand(100) * 10

trace = go.Scatter(x=x, y=y, mode='lines', name='Random Data', line=dict(color='blue'))

layout = go.Layout(
    title='Simple Line Plot',
    xaxis=dict(title='X-axis'),
    yaxis=dict(title='Y-axis'),
)

fig = go.Figure(data=[trace], layout=layout)

pyo.iplot(fig)
````
