### 44. For the given array, apply custom function to calculate moving sum along rows.


```python
import numpy as np

arr = np.array([[1, 2, 3],
                [4, 5, 6]])

def moving_sum(arr, window_size=2):
    return np.apply_along_axis(lambda x: np.convolve(x, np.ones(window_size), mode='valid'), axis=1, arr=arr)

moving_sum_result = moving_sum(arr)

print("Moving sum along rows:", moving_sum_result)
```

###  57. Write a function that extracts uppercase characters from a NumPy array of strings.


```python
import numpy as np

def extract_uppercase(arr):
    uppercase_chars = []

    for string in arr:
        for char in string:
            if char.isupper():
                uppercase_chars.append(char)

    return uppercase_chars

arr = np.array(['Hello', 'World', 'OpenAI', 'GPT'])

result = extract_uppercase(arr)
print("Uppercase characters:", result)
```

###  60. Write a function that finds the length of the longest string in a NumPy array.


```python
import numpy as np

def longest_string_length(arr):
    string_lengths = np.vectorize(len)(arr)
    return np.max(string_lengths)

arr = np.array(['apple', 'banana', 'grape', 'pineapple'])

result = longest_string_length(arr)
print("Length of the longest string:", result)
```

### 74. For the given NumPy array and swap its byte order conditionally based on system endianness using
`newbyteorder()`.


```python
import numpy as np
import sys

arr = np.array([1, 2, 3], dtype=np.int32)

native_order = '=' if sys.byteorder == 'little' else '>'

if arr.dtype.byteorder != native_order:
    swapped_array = arr.newbyteorder()
else:
    swapped_array = arr

print("Original Array:", arr)
print("Swapped Byte Order Array:", swapped_array)
```

###  85. Create a square matrix `G` and find its inverse.

```python
import numpy as np

G = np.random.randint(1, 11, size=(3, 3))

determinant_G = np.linalg.det(G)

if determinant_G != 0:
    G_inverse = np.linalg.inv(G)

    print("Matrix G:")
    print(G)
    print("\nDeterminant of Matrix G:")
    print(determinant_G)
    print("\nInverse of Matrix G:")
    print(G_inverse)
else:
    print("Matrix G is singular and cannot be inverted.")
```
