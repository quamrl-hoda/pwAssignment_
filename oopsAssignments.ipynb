### 28 Explain the use of access modifiers in Python classes.

---

**Access modifiers** in Python classes are used to control the visibility and accessibility of class attributes and methods. They help in implementing encapsulation, allowing us to restrict access to certain parts of an object’s data and methods, which can enhance the safety and maintainability of our code.

**Types of Access Modifiers in Python**
- **Public:-** Attributes and methods that are publicly accessible from outside the class.

**Example:-**

```python
class myClass:
    def __init__(self):
        self.public_attribute = "Publicaly exucatable"

    def public_method(self):
        return "This is a public method"

obj = myClass()
print(obj.public_attribute)
print(obj.public_method())
```

- **Protected:-** Attributes and methods that are intended to be protected from access outside the class and its subclasses. They can be accessed by subclasses but should not be accessed directly from outside the class hierarchy.

**Example:-**

```python
class myClass:
    def __init__(self):
        self._protected_attribute = "I am protected"

    def _protected_method(self):
        return "This is a protected method"

class subClass(myClass):
    def access_protected(self):
        return self._protected_attribute

obj = subClass()
print(obj.access_protected())
```

- **Private:-** Attributes and methods that are intended to be private to the class. They cannot be accessed directly from outside the class.

**Example:-**

```python
class myClass:
    def __init__(self):
        self.__private_attribute = "private"

    def __private_method(self):
        return "This is a private method"

    def get_private(self):
        return self.__private_attribute

obj = myClass()
print(obj.get_private())
```

### 29  What is inheritance in OOP?

---

**Inheritance** is a fundamental concept in Object-Oriented Programming (OOP) that allows a new class (called a subclass or child class) to inherit attributes and methods from an existing class (called a superclass or parent class). This mechanism enables code reusability and establishes a hierarchical relationship between classes.

**Key Features of Inheritance**

**Code Reusability:**  By inheriting from a superclass, a subclass can reuse the code defined in the superclass without having to duplicate it. This leads to more efficient and manageable code.

**Hierarchical Classification:** Inheritance allows for the creation of a class hierarchy where subclasses can specialize or extend the behavior of their parent classes. This organization makes it easier to understand relationships between different classes.

**Overriding:**  Subclasses can override methods of the superclass to provide specific implementations while retaining the same interface. This allows for polymorphic behavior, where a method can act differently based on the object that calls it.

**Multiple Inheritance:** In Python, a subclass can inherit from more than one superclass. This feature allows combining functionality from multiple sources but can lead to complexity such as the "diamond problem."

- **Example:-**

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "Some sound"

class Dog(Animal):
    def speak(self):
        return "Bark"

class Cat(Animal):
    def speak(self):
        return "Meow"

dog = Dog("Buddy")
cat = Cat("Whiskers")

print(dog.name)
print(dog.speak())

print(cat.name)
print(cat.speak())
```

### 30  Define polymorphism in OOP.

----

**Polymorphism** is one of the core concepts in Object-Oriented Programming (OOP) that allows objects of different classes to be treated as objects of a common superclass. It enables a single interface to represent different underlying forms (data types). In essence, polymorphism allows methods to do different things based on the object it is acting upon, which promotes flexibility and reusability in code.

**Types of Polymorphism :-**

**Compile-Time Polymorphism (Static Polymorphism):**

This type of polymorphism is resolved during compile time. The most common example is method overloading, where multiple methods have the same name but differ in the type or number of their parameters.

**Example:-**


```python
class Math:
    def add(self, a: int, b: int) -> int:
        return a + b

    def add(self, a: float, b: float) -> float:
        return a + b

math = Math()
print(math.add(5, 10))
print(math.add(5.5, 10.2))
```

**Run-Time Polymorphism (Dynamic Polymorphism):**

This type of polymorphism is resolved during runtime. It usually occurs through method overriding, where a subclass provides a specific implementation of a method that is already defined in its superclass.

**Example:-**


```python
class Animal:
    def speak(self):
        return "Some sound"

class Dog(Animal):
    def speak(self):
        return "Bark"

class Cat(Animal):
    def speak(self):
        return "Meow"

def animal_sound(animal: Animal):
    print(animal.speak())

dog = Dog()
cat = Cat()

animal_sound(dog)
animal_sound(cat)
```

### 37  What is abstraction in Python? How is it implemented?
---

**Abstraction** is one of the fundamental principles of Object-Oriented Programming (OOP) that focuses on hiding the complex implementation details and exposing only the necessary and relevant features of an object. It allows us to interact with an object at a higher level without needing to understand all the underlying complexities.

**Key Features of Abstraction:**

**Simplification:** Abstraction simplifies complex systems by providing a clear and simplified interface for interacting with objects.

**Hiding Complexity:** It hides the intricate details of how an object operates, allowing us to use the object without needing to know how it works internally.

**Here’s how abstraction can be implemented in Python : -**

**Abstract Classes:** An abstract class cannot be instantiated directly and can contain abstract methods that must be implemented by subclasses. We define an abstract class using the ABC class from the abc module and decorate methods with @abstractmethod.

**Concrete Subclasses:** These subclasses inherit from the abstract class and provide specific implementations for the abstract methods.

**Example :-**

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass

    @abstractmethod
    def move(self):
        pass

class Dog(Animal):
    def make_sound(self):
        return "Woof!"

    def move(self):
        return "Dog runs."

class Cat(Animal):
    def make_sound(self):
        return "Meow!"

    def move(self):
        return "Cat leaps."

def animal_sound(animal: Animal):
    print(animal.make_sound())
    print(animal.move())

dog = Dog()
cat = Cat()

animal_sound(dog)
animal_sound(cat)
```

### 39  How are abstract methods different from regular methods in Python?
---

Abstract methods and regular methods in Python have distinct characteristics that serve different purposes in the context of object-oriented programming. Here’s  the differences:

**Abstract Methods**

 **Definition**: An abstract method is a method that is declared in an abstract class but does not have any implementation in that class. It serves as a blueprint for subclasses.

 **Usage**: Abstract methods are intended to be overridden in derived classes. Subclasses must provide their own implementation of the abstract method.

**Purpose**: They define a common interface for all subclasses, ensuring that certain methods are implemented in any subclass that inherits from the abstract class. This enforces a contract for the derived classes.

**Implementation**: In Python, abstract methods are defined using the `@abstractmethod` decorator from the `abc` module. A class containing one or more abstract methods must also be declared as an abstract class by inheriting from `ABC`.

**Instantiation**: We cannot create an instance of an abstract class that contains abstract methods. Attempting to do so will raise a `TypeError`.

**Regular Methods**

**Definition**: Regular methods are defined in a class with a complete implementation. They can perform actions and return values.

**Usage**: Regular methods can be called on instances of the class, and they can be inherited by subclasses. They can also be overridden in subclasses if needed, but this is optional.

**Purpose**: They provide specific functionality and behavior for the class, allowing objects to perform actions.

**Implementation**: Regular methods are defined with a complete implementation without the need for any special decorators. They can be called directly on instances of the class.

**Instantiation**: We can create instances of a class with regular methods as long as there are no abstract methods left unimplemented in the class.





### 40  How can you achieve abstraction using interfaces in Python?

----

In Python, we can achieve abstraction using interfaces by defining abstract base classes (ABCs) that specify a set of methods that must be implemented by any concrete class that derives from them. Although Python does not have a built-in concept of interfaces like some other languages (e.g., Java), we can use the `abc` module to create abstract classes that serve as interfaces

**Steps to Achieve Abstraction Using Interfaces in Python**

**Import the `abc` Module:**  Use the `abc` module to create abstract base classes.

**Define an Abstract Base Class**: Create a class that inherits from `ABC`, the base class for defining abstract classes.

**Define Abstract Methods**: Use the `@abstractmethod` decorator to define methods that must be implemented in any subclass.

**Implement Concrete Classes**: Create subclasses that inherit from the abstract base class and implement the abstract methods.

**Example:-**




```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        """Method to make a sound."""
        pass

    @abstractmethod
    def move(self):
        """Method to describe movement."""
        pass

class Dog(Animal):
    def make_sound(self):
        return "Woof!"

    def move(self):
        return "Dog runs."

class Cat(Animal):
    def make_sound(self):
        return "Meow!"

    def move(self):
        return "Cat leaps."


def animal_action(animal: Animal):
    print(animal.make_sound())
    print(animal.move())

dog = Dog()
cat = Cat()

animal_action(dog)
animal_action(cat)
```

### 41 Can you provide an example of how abstraction can be utilized to create a common interface for a group of related classes in Python?

---

```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):
    @abstractmethod
    def area(self):
        """Calculate the area of the shape."""
        pass

    @abstractmethod
    def perimeter(self):
        """Calculate the perimeter of the shape."""
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * (self.radius ** 2)

    def perimeter(self):
        return 2 * math.pi * self.radius

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)

def print_shape_details(shape: Shape):
    print(f"Area: {shape.area():.2f}")
    print(f"Perimeter: {shape.perimeter():.2f}")

circle = Circle(5)
rectangle = Rectangle(4, 6)

print("Circle Details:")
print_shape_details(circle)
print("\nRectangle Details:")
print_shape_details(rectangle)
```

- **`__iter__()`**: This method is responsible for returning an iterator object. It is required for making an object iterable. When an object is used in a `for` loop or with the `iter()` function, Python calls the `__iter__()` method to get the iterator.
  
- **`__next__()`**: This method is responsible for returning the next element of the iterable. When the `next()` function is called, Python internally calls `__next__()` to get the next item. If there are no more items to return, it raises the `StopIteration` exception to signal the end of iteration.

### **Iterator Protocol**:
- An **iterable** is an object that can be iterated over (e.g., lists, strings, dictionaries).
- An **iterator** is an object that produces a stream of data, one element at a time, using the `__next__()` method.



### Example



```python
class CountDown:
    def __init__(self, start):
        self.start = start
        self.current = start


    def __iter__(self):
        return self

    def __next__(self):
        if self.current <= 0:
            raise StopIteration
            self.current -= 1
            return self.current

countdown = CountDown(5)

for num in countdown:
    print(num)
```

### 71 How does encapsulation relate to property decorators in Python? Provide an example showcasing encapsulation using property decorators.

---

### **Encapsulation and Property Decorators in Python**

**Encapsulation** is one of the key principles of Object-Oriented Programming (OOP). It refers to the bundling of data (attributes) and the methods (functions) that operate on that data into a single unit, or class. Additionally, encapsulation involves restricting direct access to certain attributes and allowing access to them only through well-defined interfaces (methods), which helps protect the internal state of an object and ensures controlled access.

In Python, **property decorators** (`@property`, `@setter`, `@deleter`) enable encapsulation by providing a way to control access to an object's attributes. They allow you to hide the internal representation of the data and provide controlled access through getter and setter methods. This means that you can:
- Control how an attribute is accessed or modified.
- Prevent external code from directly modifying an attribute in an invalid or uncontrolled manner.
- Add logic or validation to ensure the integrity of the data.

### **How Property Decorators Enable Encapsulation**:
- **`@property`**: Allows a method to be accessed like an attribute, but behind the scenes, you can compute or validate the value.
- **`@setter`**: Controls how an attribute's value is set, ensuring that only valid data is assigned.
- **`@deleter`**: Controls how an attribute is deleted, allowing for additional cleanup or restrictions.

### Example



```python
from datetime import datetime

class Person:
    def __init__(self, name, birth_year):
        self._name = name
        self._birth_year = birth_year
    @property
    def age(self):
        current_year = datetime.now().year
        return current_year - self._birth_year

    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("Age cannot be negative.")
        current_year = datetime.now().year
        self._birth_year = current_year - value


    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, value):
        if not value:
            raise ValueError("Name cannot be empty.")
        self._name = value

person = Person("quamrul", 1990)

print(f"{person.name} is {person.age} years old.")

person.age = 35
print(f"After update, {person.name} is {person.age} years old (Birth Year: {2024 - person.age}).")

try:
    person.age = -5
except ValueError as e:
    print(e)
```
